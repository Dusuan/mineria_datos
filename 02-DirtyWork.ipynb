```python
import random

random_list = [random.random() for _ in range(5_000_000)]

print("Random list length:" , len(random_list))
```

```python
def custom_standard_deviation(list):
    n = 0
    eqtop = 0
    pop_mean = 0

    for i in random_list:
        n += 1

    for x in random_list:
        pop_mean += x

    pop_mean /= n

    for x in random_list:
        eqtop += (x - pop_mean) ** 2

    return (eqtop / n)**0.5
```

```python
print("Standard deviation:", custom_standard_deviation(random_list))
```
